### Ejercicio 1: Área de un triángulo

- Se aplica el modelo **Entrada-Proceso-Salida**: la entrada son la `base` y la `altura` capturadas con `input()` y convertidas con `float()`.
- El **proceso** calcula el área con la fórmula `area = (base * altura) / 2`.
- La **salida** muestra el resultado con `print()`.



```python
base = float(input("Base del triángulo: "))
altura = float(input("Altura del triángulo: "))
area = (base * altura) / 2
print(f"El área del triángulo es: {area}")
```

### Ejercicio 5: Casting tras `input()`

- `input()` siempre devuelve una cadena (`str`); se demuestra con `type()`.
- Se intenta `x + 1` dentro de `try-except` capturando el `TypeError`, porque sumar una cadena y un entero no es válido.
- Finalmente se convierte con `int(x)` para poder sumar y obtener el resultado correcto.



```python
x = input("Ingresa un número: ")
print("x es tipo", type(x).__name__)
try:
    x + 1
except TypeError:
    print("x + 1 genera TypeError")
print("int(x) + 1 =", int(x) + 1)
```

### Ejercicio 10: Suma acumulada con `while`

- Se piden números con `input()` mientras sean distintos de `0`.
- El `0` **termina el ciclo** y no se suma; cada número distinto de cero se acumula en `suma`.



```python
suma = 0
numero = int(input("Ingresa un número (0 para terminar): "))
while numero != 0:
    suma += numero
    numero = int(input("Ingresa un número (0 para terminar): "))
print(f"La suma de los números es: {suma}")
```

### Ejercicio 12: Suma de pares del 1 al 20 con `for` y `continue`

- Un ciclo `for` recorre los números del 1 al 20.
- Con `continue` se saltan los impares; únicamente los números pares se acumulan en `suma`.



```python
suma = 0
for numero in range(1, 21):
    if numero % 2 != 0:
        continue
    suma += numero
print(f"La suma de los números pares del 1 al 20 es: {suma}")
```

### Ejercicio 13: Suma de una lista con función que regresa valor

- `suma_lista(lista)` **regresa valor** (con `return`): recorre con un ciclo `for` los elementos de la lista y acumula la suma, sin usar `sum()`.
- Se aplica a la lista `[10, 20, 30, 40]`.



```python
def suma_lista(lista):
    total = 0
    for elemento in lista:
        total += elemento
    return total


numeros = [10, 20, 30, 40]
print(f"La suma de los elementos es: {suma_lista(numeros)}")
```

### Ejercicio 14: Máximo de una lista con función que regresa valor

- `maximo_lista(lista)` **regresa valor** (con `return`): inicializa `maximo` con el primer elemento y lo actualiza al compararlo con el resto, sin usar `max()`.
- Se aplica a la lista `[7, 3, 9, 2]`.



```python
def maximo_lista(lista):
    maximo = lista[0]
    for elemento in lista:
        if elemento > maximo:
            maximo = elemento
    return maximo


numeros = [7, 3, 9, 2]
print(f"El elemento máximo es: {maximo_lista(numeros)}")
```

### Ejercicio 15: Matriz y suma de la diagonal principal

- `suma_diagonal(matriz)` **regresa valor** (con `return`): con un ciclo `for` sobre los índices suma los elementos donde el renglón coincide con la columna (`matriz[i][i]`).
- Se aplica a la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9]]`.


```python
def suma_diagonal(matriz):
    suma = 0
    for i in range(len(matriz)):
        suma += matriz[i][i]
    return suma


matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(f"La suma de la diagonal principal es: {suma_diagonal(matriz)}")
```

### Ejercicio 16: Promedio con función que no regresa valor

- `promedio(valores)` **no regresa valor**: acumula la suma con un ciclo `for` e imprime el promedio formateado con dos decimales (`:.2f`) dentro de la propia función.
- Se aplica a la lista `[8, 9, 10, 7]`.



```python
def promedio(valores):
    suma = 0
    for valor in valores:
        suma += valor
    print(f"El promedio es: {suma / len(valores):.2f}")


calificaciones = [8, 9, 10, 7]
promedio(calificaciones)
```

### Ejercicio 18: Factorial con función recursiva

- `factorial(n)` **regresa valor** (con `return`): tiene caso base cuando `n` es `0` o `1`, y en cualquier otro caso se llama a sí misma con `n - 1` (recursión).
- Se captura `n` con `input()`.



```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


n = int(input("Número: "))
print(f"El factorial de {n} es: {factorial(n)}")
```

### Ejercicio 19: Tupla: acceso y suma con función

- Se accede al tercer elemento de la tupla con el índice `2`.
- `suma_tupla(tupla)` **regresa valor** (con `return`): acumula la suma de los elementos con un ciclo `for`.



```python
def suma_tupla(tupla):
    total = 0
    for elemento in tupla:
        total += elemento
    return total


tupla = (4, 8, 15, 16, 23, 42)
print(f"El tercer elemento es: {tupla[2]}")
print(f"La suma de la tupla es: {suma_tupla(tupla)}")
```

### Ejercicio 20: Diccionario: buscar teléfono de un contacto

- La agenda es un diccionario cuya **clave** es el nombre del contacto y su **valor** el teléfono.
- `buscar_telefono(agenda, nombre)` **regresa valor** (con `return`): consulta el teléfono con `get()`.
- Se captura el nombre a buscar con `input()`.



```python
def buscar_telefono(agenda, nombre):
    return agenda.get(nombre)


agenda = {
    "Ana": "5551234",
    "Beto": "5555678",
    "Carolina": "5551111"
}
nombre = input("Nombre del contacto a buscar: ")
print(f"El teléfono de {nombre} es: {buscar_telefono(agenda, nombre)}")
```

### Ejercicio 21: Diccionario de calificaciones y promedios

- El diccionario asocia cada **nombre** con su **lista de calificaciones**.
- Se recorre con `.items()` y un ciclo `for` calcula el promedio de cada alumno, formateado a dos decimales.



```python
def promedio(notas):
    total = 0
    for nota in notas:
        total += nota
    return total / len(notas)


calificaciones = {"Ana": [8, 9, 10], "Beto": [7, 7, 8]}
for nombre, notas in calificaciones.items():
    print(f"{nombre}: {promedio(notas):.2f}")
```

### Ejercicio 24: Contar palabras de un string

- `contar_palabras(mensaje)` **regresa valor** (con `return`): usa `split()` para separar el mensaje en palabras y `len()` para contarlas.
