# Extra 1: Sistema de calificaciones con tuplas
**Planteamiento del problema:** Se tiene una tupla inicial de calificaciones y se necesita mostrar la tercera calificación, capturar dos calificaciones nuevas con input() y anexarlas en una nueva tupla, convertir esa tupla en lista para ordenarla de mayor a menor, y calcular la suma de todos los elementos mediante una función que retorne el valor. Se elige la tupla porque sus elementos no deben modificarse; la lista permite el ordenamiento y la función permite reutilizar el cálculo de la suma.

```python
def suma_calificaciones(calificaciones):
    total = 0
    for c in calificaciones:
        total += c
    return total

calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)
print("Tercera calificación:", calificaciones[2])
nueva1 = float(input("Nueva calificación 1: "))
nueva2 = float(input("Nueva calificación 2: "))
nueva_tupla = calificaciones + (nueva1, nueva2)
print("Nueva tupla:", nueva_tupla)
lista_ordenada = sorted(list(nueva_tupla), reverse=True)
print("Lista ordenada (mayor a menor):", lista_ordenada)
print("Suma total:", suma_calificaciones(nueva_tupla))
```

# Extra 2: Agenda de contactos con búsqueda
**Planteamiento del problema:** Se necesita una agenda con tres contactos iniciales, capturar un nuevo contacto (nombre y teléfono) con input() y agregarlo, imprimir todos los nombres con keys() en una sola línea, y buscar el teléfono del nuevo contacto mediante una función que retorne el valor. Se elige el diccionario porque la clave (nombre) está asociada directamente al valor (teléfono) y permite una búsqueda eficiente.

```python
def buscar_telefono(agenda, nombre):
    return agenda.get(nombre)

agenda = {
    "Montserrat": "4411676121",
    "Dulce": "5561524514",
    "Carolina": "3312456789"
}
nombre_nuevo = input("Nombre del nuevo contacto: ")
telefono_nuevo = input("Teléfono del nuevo contacto: ")
agenda[nombre_nuevo] = telefono_nuevo
print("Contactos registrados:", ", ".join(agenda.keys()))
busqueda = input("Nombre a buscar: ")
telefono = buscar_telefono(agenda, busqueda)
print("El teléfono de", busqueda, "es:", telefono)
```

# Extra 3: Calculadora segura con manejo de excepciones
**Planteamiento del problema:** El programa debe pedir dos números enteros y realizar la división del primero entre el segundo. Si el usuario ingresa caracteres no numéricos o vacíos se genera un ValueError que debe capturarse con un mensaje controlado, y si el segundo número es cero se debe capturar el ZeroDivisionError con un mensaje amigable. El bloque try-except garantiza que el programa no se detenga de forma abrupta ante errores de entrada o de cálculo.

```python
try:
    num1 = int(input("Primer número entero: "))
    num2 = int(input("Segundo número entero: "))
    resultado = num1 / num2
    print(num1, "dividido entre", num2, "es:", resultado)
except ValueError:
    print("Error, Debes ingresar caracteres numéricos. No se aceptan valores vacíos ni no numéricos.")
except ZeroDivisionError:
    print("Error, No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
```

# Extra 4: Analizador de mensajes con strings
**Planteamiento del problema:** Se tiene un mensaje fijo y se necesita obtener su longitud con len(), transformarlo a mayúsculas, reemplazar la palabra "Flamewall" por "Fairy Knife Hell" y contar cuántas palabras contiene mediante una función que retorne el conteo. Se utilizan los métodos de strings (upper, replace, split) porque simplifican la transformación del texto y el conteo de palabras del mensaje original.

```python
def contar_palabras(mensaje):
    return len(mensaje.split())

mensaje = "Mi canción favorita es Flamewall, es también mi nivel favorito de Geometry Dash"
print("Longitud del mensaje:", len(mensaje))
print("En mayúsculas:", mensaje.upper())
print("Texto reemplazado:", mensaje.replace("Flamewall", "Fairy Knife Hell"))
print("Palabras totales:", contar_palabras(mensaje))
```
